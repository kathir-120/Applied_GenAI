

# While **Loops**


```python
# intializition ,condition,increment/decrement
#print values from 1 to 20
c=1
while c<=20:
  print(c)
  c+=1
```

```python
i = 20

while i >= 1:
    print(i)
    i -= 1
```


For Loops


```python
  #create loop
arr = ["apple","banana","cherry"]
for i in arr:
    print(i)
```

```python
for i in range(5,50,5):
  print(i)
```
