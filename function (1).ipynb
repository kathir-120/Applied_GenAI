```python
#addition of two numbers

def additionTwoNumbers(a,b):
  print('user give value',a)
  print('user give value',b)
  c = a+b
  print('total value of',c)
```

```python
additionTwoNumbers(1,2) # we are called as positional aruguments
```

```python
additionTwoNumbers(b=1,a=2) # we are called as keyword aruguments
```

```python
# default parameter of b = 10

def additionTwoNumbers(a,b=10):
  print('user give value',a)
  print('user give value',b)
  c = a+b
  print('total value of',c)
```

```python
additionTwoNumbers(20)
```

```python
additionTwoNumbers(20,30) # 40 overrides default argument of 10
```

```python
# default parameter of b = 10

def additionTwoNumbers(a,b=10):
  print('user give value',a)
  print('user give value',b)
  c = a+b
  return 'total value of',c
```

```python
print(additionTwoNumbers(20,30))
```
