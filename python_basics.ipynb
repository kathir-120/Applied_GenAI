# **Python - Slicing Strings**

**#Important: Strings are immutable in Python, so returns a new string; it does not modify the original string.**

 **string is immutable because we cannot change its original value after it is created.**

```python
a = "Hello world"
print(a[1:4]);  # Python slicing works from left to right when the step value is positive.
print(a[-1::-1]); # Python slicing works from right to left when the step value is negative.
```
