# **Tuple Data Types**

**Tuple three Condition**

Ordered - Items keep their order,
Unchangeable - Data cannot be changed directly,
Duplicate allowed - Repeated values are allowed

```python
# Python automatically creates a tuple when you separate multiple values with commas, even without parentheses ().
# Tuples are unchangeable, meaning that we cannot change, add or remove items after the tuple has been created.
a = 1, 2, 3, 4, 4
print(a)
```

```python
# If the number of variables is less than the number of values, you can add an * to the variable name and the values will be assigned to the variable as a list.
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
(green, yellow, *red) = fruits;
print(green);
print(yellow);
print(red);
```

# **Set Data Types**

**Unordered: Items do not have a fixed position**,
**No duplicates: Duplicate values are automatically remove,**
**Mutable: You can add or remove items,**
**Unindexed: You cannot access items using an index like fruits[0],**

```python
myset = {"apple", "banana", "cherry"}
print(myset);
# Unordered means that the items in a set do not have a defined order.
```

```python
# “Set items are unchangeable” means you cannot directly change an existing item inside a set.
fruits = {"apple", "banana", "cherry"}
fruits[0] = "orange"  # Error
```

```python
fruits = {"apple", "banana", "cherry"};
fruits.add("orange");
fruits.remove("banana");
print(fruits);
# However,a set itself is mutable — you can add or remove items.
```

```python
thisset = {"apple", "banana", "cherry", "apple"}
print(thisset);
# Yes! ✅ In Python, a Set automatically removes duplicate values.
```

```python
thisset = {"apple", "banana", "cherry"};
tropical = {"pineapple", "mango", "papaya"};
thisset.update(tropical);
print(thisset);
```

```python
fruits = {"apple", "cherry"};
fruits.remove("banana");
print(fruits);
# remove() → Item illana error varum.
```

```python
thisset = {"apple", "cherry"};
thisset.discard("banana");
print(thisset);
# - discard() → Item illana kooda error varaadhu.
```

```python
# update() method is used to add multiple elements from another iterable into a set.
fruits = {"apple", "banana"};
more_fruits = {"orange", "mango"};
fruits.update(more_fruits);
print(fruits);
```

```python
# The union() method returns a new set with all items from both sets.

set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = set1.union(set2)
print(set3)
```

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = set1 | set2
print(set3)

# join multiple set

set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"John", "Elena"}
set4 = {"apple", "bananas", "cherry"}
myset = set1 | set2 | set3 |set4
print(myset)
```

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"John", "Elena"}
set4 = {"apple", "bananas", "cherry"}
myset = set1.union(set2, set3, set4)
print(myset)
```

```python
# intersection() is a Python set method that returns a new set containing only the elements common to both sets. It does not modify the original sets.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set3 = set1.intersection(set2)
print(set3)

# You can use the & operator instead of the intersection() method, and you will get the same result.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set3 = set1 & set2
print(set3)
```

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set1.intersection_update(set2)
print(set1);
# The intersection_update() method will also keep ONLY the duplicates, but it will change the original set instead of returning a new set.
```

```python
# Returns elements present in the first set but not the second
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}
z = x.difference(y)
print(z)
```

```python
# The difference_update() method will keep the items from the first set that are not in the other set, but it will change the original set instead of returning a new set.
set1 = {"apple", "banana", "cherry"};
set2 = {"google", "microsoft", "apple"};
set1.difference_update(set2);
print(set1);
```

```python
# The symmetric_difference() method will keep only the elements that are NOT present in both sets.

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"};
set3 = set1.symmetric_difference(set2);
print(set3);
```

```python
# The symmetric_difference_update() method will also keep all but the duplicates, but it will change the original set instead of returning a new set.
set1 = {"apple", "banana", "cherry"};
set2 = {"google", "microsoft", "apple"};
set1.symmetric_difference_update(set2);
print(set1)
```

```python
# Shallow copy means creating a new set with the same elements as the original set.
a = {10, 20, 30}
b = a.copy()
print(b)
print(a == b)
print(a is b)
```

```python
# Meaning: A shallow copy creates a new outer list, but the inner lists are shared.
import copy
a = [[10, 20], [30, 40]];
b = copy.copy(a);
b[0][0] = 100;
print(a);
print(b);
```

```python
# Meaning: A deep copy creates a new outer list and copies the inner lists too.
import copy
a = [[10, 20], [30, 40]]
b = copy.deepcopy(a)
b[0][0] = 100
print("Original:", a)
print("Copy:", b)
```
