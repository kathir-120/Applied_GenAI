***if elif statement***

```python
#Creating Grading System
# >90 'A',75 to 90 'B',60 to 70 'C',35 to 60 'D', 20 to 35 'E',0 to 20 'F'

score = 85
if(score>=90):
  print('A')
elif(score>=75) and (score <90):
  print('B')
elif(score>=60) and (score <75):
  print('C')
elif(score>=35)and (score <60):
  print('D')
elif(score>=20)and (score <35):
  print('E')
else:
  print("F")
```
