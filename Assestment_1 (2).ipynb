```python
# 2.Create a list and perform basic operations like append(), remove(), pop(), and slicing.
# Iterate over a list using loops.

fruits = ['apple', 'banana', 'cherry','mango']

# Access the list elements
print(fruits[0])

#Modifying list elements
fruits[0] = 'orange'
print(fruits)

#Slicing the list elements
list_fruits = fruits[0:2]
print(list_fruits)

# Using negative value for slicing in list elements
list_fruits = fruits[-3:-1]
print(list_fruits)

# List slicing with postive step values
list_fruits = fruits[0::2]
print(list_fruits)

# List slicing with negative step values
list_fruits = fruits[-1::-2]
print(list_fruits)

#Using the append() function add the item the end of the list
fruits = ['apple', 'banana', 'cherry','mango']
fruits.append('orange')
fruits.append(('kiwi','banana'))
fruits.append({'grap','Pineapple'})
fruits.append({'fruitsname':'dragonfruit'})
print(fruits)

#Using insert() function add the items in the list elements
fruits = ['apple', 'banana', 'cherry','mango']
fruits.insert(0,'orange')
fruits.insert(1,'jackfruit')
print(fruits)


#Using remove() function delete from the list elements
fruits = ['apple', 'banana', 'cherry','mango']
fruits.remove('apple')
print(fruits)

#Using pop() function remove the list elements
fruits = ['apple', 'banana', 'cherry','mango']
fruits.pop()
print(fruits)

fruits.pop(1)
print(fruits)

#Calculate the list length using the len() function
fruits = ['apple', 'banana', 'cherry','mango']
print(len(fruits))

#Reversing a List Using reverse() Method it was change values in original list.
numbers = [10, 20, 30, 40, 50]
numbers.reverse()
print(numbers)

#Sorting a List Using sort() Method
#Ascending Order
numbers = [50,40,30,20,10]
numbers.sort()
print(numbers)

#Descending Order
numbers = [10,20,30,40,50]
numbers.sort(reverse=True)
print(numbers)

#Using Another sorted() method it was not affect original list values.
#Ascending Order
numbers = [50,40,30,20,10]
print(sorted(numbers))
print(numbers)

#Descending Order
numbers = [10,20,30,40,50]
print(sorted(numbers,reverse=True))
print(numbers)


#Using index() function find the element position in the list elements
numbers = [10,20,30,40,50]
print(numbers.index(30))

#Using the count() function find the element how many times came from the list
numbers = [10,20,30,40,50,10,20,30,40,50]
print(numbers.count(10))

#Using calculate the max and min and sum
print(max(numbers))
print(min(numbers))
print(sum(numbers))

#Using for loop iterate the list
fruits = ['apple', 'banana', 'cherry','mango']
for x in fruits:
  print(x)

#Using the range() function
fruits = ['apple', 'banana', 'cherry','mango']
for x in range(len(fruits)):
  print(fruits[x])

#Using while loop iterate the list
fruits = ['apple', 'banana', 'cherry','mango']
i = 0
while i < len(fruits):
  print(fruits[i])
  i = i + 1

```

```python
# Set Operations:
# Create a set and perform operations like union, intersection, and difference.

#Using union() function in set
a = {1,2,3,4}
b = {3,4,5,6}
# The union() function is used to combine two or more sets into a single set.
print(a.union(b))
print(a|b)

# The intersection() function is used to find the common elements between two or more sets.
print(a.intersection(b))
print(a&b)

# The difference() function is used to find the elements that are in the first set but not in the second set.
print(a.difference(b))
print(a-b)
print(b-a)

# Use ^ operator for symmetric difference
# The symmetric_difference() function is used to find elements that are present in either of two sets, but not in both sets.
E = {0, 2, 4, 6, 8};
N = {1, 2, 3, 4, 5};
print(E ^ N)



# Understand the uniqueness property of sets.
fruits = {"apple", "banana", "cherry"}
print(fruits);

#Using add() function in set
fruits.add("orange")
print(fruits)

#Using the update() function in set
fruits.update(["jack fruits","green apple"])
fruits.update({"mango","pineapple"})
fruits.update(("kiwi","banana"))
print(fruits)

#Using remove() function remove the set items
fruits.remove("kiwi")
print(fruits)

#Using discard() function remove the items
fruits.discard("banana")
print(fruits)

#Using pop() function remove the set items
fruits.pop()
print(fruits)

#Using clear() function remove the all set items
print(fruits.clear())

```
